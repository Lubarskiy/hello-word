# Урок 01. Старт, ввод/вывод, первые функции

## Мини-теория
- Пишем решения через функции — так удобнее тестировать и переносить в VS Code.
- Форматирование строк через f-строки.
- Краевые случаи: деление на ноль, пустая строка.
- Дипломный артефакт: «паспорт проекта» как dict.

## Практика
> В этой версии вставлено **готовое решение**. Можно удалить решения и оставить TODO для студентов.


```python

# =========================
# LESSON 01 — SOLUTION
# =========================

def w1_print_hello_and_name(name: str) -> None:
    print("Hello, AI Diploma!")
    print(name)


def p1_format_person(name: str, age: int) -> str:
    return f"Имя: {name}, Возраст: {age}"


def p2_basic_ops(a: float, b: float):
    s = a + b
    d = a - b
    p = a * b
    q = None if b == 0 else round(a / b, 2)
    return (s, d, p, q)


def p3_string_edges(s: str):
    if s == "":
        return (0, None, None)
    return (len(s), s[0], s[-1])


def p4_min_max3(a: float, b: float, c: float):
    return (min(a, b, c), max(a, b, c))


def d1_project_passport(title: str, goal: str, user: str, risks: list[str]) -> dict:
    return {"title": title, "goal": goal, "user": user, "risks": risks}


def run_all_tests_lesson_01():
    assert p1_format_person("Ivan", 20) == "Имя: Ivan, Возраст: 20"
    assert p1_format_person("Сергей", 55) == "Имя: Сергей, Возраст: 55"

    assert p2_basic_ops(10, 2) == (12, 8, 20, 5.0)
    assert p2_basic_ops(10, 3) == (13, 7, 30, 3.33)
    assert p2_basic_ops(10, 0) == (10, 10, 0, None)

    assert p3_string_edges("abc") == (3, "a", "c")
    assert p3_string_edges("x") == (1, "x", "x")
    assert p3_string_edges("") == (0, None, None)

    assert p4_min_max3(1, 2, 3) == (1, 3)
    assert p4_min_max3(3, 2, 1) == (1, 3)
    assert p4_min_max3(-5, -5, 0) == (-5, 0)

    pp = d1_project_passport("AI Helper", "Помогать пользователю", "HR", ["data quality", "privacy"])
    assert isinstance(pp, dict)
    assert pp["title"] == "AI Helper"
    assert pp["goal"] == "Помогать пользователю"
    assert pp["user"] == "HR"
    assert pp["risks"] == ["data quality", "privacy"]

    print("✅ LESSON 01: all tests passed")
```

### Lesson 01 Functions Explanation:

#### `w1_print_hello_and_name(name: str) -> None`
This function takes a `name` as a string input. It prints a greeting "Hello, AI Diploma!" followed by the provided `name` on a new line. It doesn't return any value.

#### `p1_format_person(name: str, age: int) -> str`
This function takes a person's `name` (string) and `age` (integer). It uses an f-string to format and return a string in the format "Имя: {name}, Возраст: {age}".

#### `p2_basic_ops(a: float, b: float)`
This function performs basic arithmetic operations (sum, difference, product, and quotient) on two float numbers, `a` and `b`.
- It calculates the sum (`s`), difference (`d`), and product (`p`).
- For the quotient (`q`), it checks if `b` is zero to prevent `ZeroDivisionError`. If `b` is zero, `q` is `None`; otherwise, it calculates `a / b` and rounds it to two decimal places.
- It returns a tuple containing `(s, d, p, q)`.

#### `p3_string_edges(s: str)`
This function processes a string `s`.
- If the string is empty (`""`), it returns a tuple `(0, None, None)` (length, first character, last character).
- Otherwise, it returns a tuple containing the length of the string, its first character, and its last character.

#### `p4_min_max3(a: float, b: float, c: float)`
This function takes three float numbers (`a`, `b`, `c`). It returns a tuple containing the minimum and maximum of these three numbers.

#### `d1_project_passport(title: str, goal: str, user: str, risks: list[str]) -> dict`
This function creates a dictionary representing a "project passport". It takes the project's `title`, `goal`, `user`, and a list of `risks` as input. It returns a dictionary with these keys and their corresponding values.

#### `run_all_tests_lesson_01()`
This function serves as a test suite for all the other functions in Lesson 01. It contains a series of `assert` statements that check if the output of each function for specific inputs matches the expected results. If all assertions pass, it prints `✅ LESSON 01: all tests passed`.

# Урок 02. Окружение и утилиты (байты, среднее, env-check)

## Мини-теория
- Версия Python и утилиты — часть инженерной культуры.
- Парсинг чисел должен быть безопасным (try/except).
- В дипломе важно уметь проверять окружение.

## Практика
> В этой версии вставлено **готовое решение**. Можно удалить решения и оставить TODO для студентов.


```python

# =========================
# LESSON 02 — SOLUTION
# =========================

import sys


def w1_python_version() -> str:
    v = sys.version_info
    return f"{v.major}.{v.minor}.{v.micro}"


def p1_format_bytes(n: int) -> str:
    if n < 0:
        raise ValueError("n must be >= 0")

    units = ["B", "KB", "MB", "GB", "TB"]
    value = float(n)
    unit_idx = 0

    while value >= 1024 and unit_idx < len(units) - 1:
        value /= 1024.0
        unit_idx += 1

    unit = units[unit_idx]
    if unit == "B":
        return f"{int(value)} {unit}"
    return f"{value:.2f} {unit}"


def p2_mean(nums: list[float]):
    if not nums:
        return None
    return sum(nums) / len(nums)


def p3_summary(nums: list[float]) -> dict:
    if not nums:
        return {"min": None, "max": None, "avg": None}
    return {"min": min(nums), "max": max(nums), "avg": p2_mean(nums)}


def p4_parse_floats_safe(values: list[str]) -> list[float]:
    out: list[float] = []
    for s in values:
        try:
            out.append(float(s))
        except (TypeError, ValueError):
            continue
    return out


def d1_env_check(major: int, minor: int) -> dict:
    ok = (major, minor) >= (3, 10)
    msg = "Python environment OK" if ok else "Python version too old: need Python >= 3.10"
    return {"ok": ok, "python_major": major, "python_minor": minor, "message": msg}


def run_all_tests_lesson_02():
    v = w1_python_version()
    assert isinstance(v, str) and len(v.split(".")) >= 2

    assert p1_format_bytes(0) == "0 B"
    assert p1_format_bytes(999) == "999 B"
    assert p1_format_bytes(1024) == "1.00 KB"
    assert p1_format_bytes(1536) == "1.50 KB"
    assert p1_format_bytes(1024 * 1024) == "1.00 MB"

    assert p2_mean([1, 2, 3]) == 2.0
    assert p2_mean([]) is None

    assert p3_summary([1, 2, 3]) == {"min": 1, "max": 3, "avg": 2.0}
    assert p3_summary([]) == {"min": None, "max": None, "avg": None}

    assert p4_parse_floats_safe(["1", "2.5", "x", "3"]) == [1.0, 2.5, 3.0]
    assert p4_parse_floats_safe(["bad", ""]) == []

    assert d1_env_check(3, 10)["ok"] is True
    assert d1_env_check(3, 9)["ok"] is False
    assert "Python" in d1_env_check(3, 9)["message"]

    print("✅ LESSON 02: all tests passed")
```

```python
run_all_tests_lesson_02()
```

# Урок 03. Типы и safe-конвертеры, нормализация ввода

## Мини-теория
- Данные часто «грязные» → нужны safe-конвертеры.
- Нормализация: привести типы и убрать пробелы.
- Soft-parse: возвращаем ошибки списком, не падаем.

## Практика
> В этой версии вставлено **готовое решение**. Можно удалить решения и оставить TODO для студентов.


```python

# =========================
# LESSON 03 — SOLUTION
# =========================

def p1_to_int_safe(s: str, default: int) -> int:
    try:
        return int(str(s).strip())
    except (TypeError, ValueError):
        return default


def p2_to_float_safe(s: str, default: float) -> float:
    try:
        return float(str(s).strip())
    except (TypeError, ValueError):
        return default


def p3_detect_number_type(s: str) -> str:
    ss = str(s).strip()
    if ss == "":
        return "other"

    try:
        iv = int(ss)
        if str(iv) == ss:
            return "int"
    except ValueError:
        pass

    try:
        float(ss)
        if "." in ss:
            return "float"
    except ValueError:
        pass

    return "other"


def p4_is_email_like(s: str) -> bool:
    if not isinstance(s, str):
        return False
    if s.count("@") != 1:
        return False
    left, right = s.split("@")
    if left == "" or right == "":
        return False
    if "." not in right:
        return False
    if right.startswith(".") or right.endswith("."):
        return False
    return True


def p5_normalize_user_input(d: dict) -> dict:
    raw_id = d.get("id")
    raw_title = d.get("title", "")
    raw_score = d.get("score")

    title = "" if raw_title is None else str(raw_title).strip()

    id_val = None
    if raw_id is not None:
        try:
            id_val = int(str(raw_id).strip())
        except (TypeError, ValueError):
            id_val = None

    score_val = None
    if raw_score is not None:
        try:
            score_val = float(str(raw_score).strip())
        except (TypeError, ValueError):
            score_val = None

    return {"id": id_val, "title": title, "score": score_val}


def d1_parse_item_soft(raw: dict):
    errors: list[str] = []
    norm = p5_normalize_user_input(raw)

    if norm["title"] == "":
        errors.append("title: required non-empty string")

    if raw.get("id") is not None:
        if norm["id"] is None:
            errors.append("id: must be integer")
        elif norm["id"] < 0:
            errors.append("id: must be >= 0")

    if raw.get("score") is not None:
        if norm["score"] is None:
            errors.append("score: must be float")
        elif not (0.0 <= norm["score"] <= 1.0):
            errors.append("score: must be in [0, 1]")

    return norm, errors


def run_all_tests_lesson_03():
    assert p1_to_int_safe("42", 0) == 42
    assert p1_to_int_safe("x", 7) == 7
    assert p1_to_int_safe("  10 ", 0) == 10

    assert p2_to_float_safe("3.14", 0.0) == 3.14
    assert p2_to_float_safe("x", 1.5) == 1.5
    assert p2_to_float_safe(" 2 ", 0.0) == 2.0

    assert p3_detect_number_type("10") == "int"
    assert p3_detect_number_type("-12") == "int"
    assert p3_detect_number_type("3.14") == "float"
    assert p3_detect_number_type("-0.5") == "float"
    assert p3_detect_number_type("x") == "other"

    assert p4_is_email_like("a@b.com") is True
    assert p4_is_email_like("ab.com") is False
    assert p4_is_email_like("a@@b.com") is False
    assert p4_is_email_like("@b.com") is False
    assert p4_is_email_like("a@bcom") is False

    out = p5_normalize_user_input({"id": "5", "title": "  Test  ", "score": "0.8"})
    assert out == {"id": 5, "title": "Test", "score": 0.8}
    out = p5_normalize_user_input({"id": "x", "title": "  ", "score": "no"})
    assert out["id"] is None and out["score"] is None and out["title"] == ""

    item, errors = d1_parse_item_soft({"id": "3", "title": "  A  ", "score": "0.9"})
    assert errors == []
    assert item["id"] == 3 and item["title"] == "A" and item["score"] == 0.9

    item, errors = d1_parse_item_soft({"id": "-1", "title": "X", "score": "2"})
    assert "id" in " ".join(errors).lower()
    assert "score" in " ".join(errors).lower()

    item, errors = d1_parse_item_soft({"title": "   "})
    assert len(errors) >= 1

    print("✅ LESSON 03: all tests passed")
```

```python
run_all_tests_lesson_03()
```

# Урок 04. Условия и валидаторы (username + entity validator)

## Мини-теория
- Валидатор возвращает список ошибок, не падает.
- Проверяем крайние случаи: пустые строки, неверные типы.
- Переносится в диплом как слой validators.

## Практика
> В этой версии вставлено **готовое решение**. Можно удалить решения и оставить TODO для студентов.


```python

# =========================
# LESSON 04 — SOLUTION
# =========================

def p1_even_odd(x: int) -> str:
    return "even" if x % 2 == 0 else "odd"


def p2_temp_label(t: float) -> str:
    if t < 10:
        return "cold"
    if t <= 25:
        return "ok"
    return "hot"


def p3_tariff_cost(requests_count: int, price_per_req: float) -> float:
    if requests_count <= 10:
        return 0.0
    return float(requests_count - 10) * float(price_per_req)


def p4_clamp(x: float, lo: float, hi: float) -> float:
    if x < lo:
        return lo
    if x > hi:
        return hi
    return x


def p5_validate_username(username: str) -> list[str]:
    errors: list[str] = []
    if not isinstance(username, str):
        return ["username: must be a string"]

    if not (3 <= len(username) <= 20):
        errors.append("username: length must be 3..20")

    allowed = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789_"
    for ch in username:
        if ch not in allowed:
            errors.append("username: only letters, digits and '_' allowed")
            break
    return errors


def d1_validate_item(item: dict) -> list[str]:
    errors: list[str] = []
    if not isinstance(item, dict):
        return ["item: must be dict"]

    title = item.get("title")
    if not isinstance(title, str) or title.strip() == "":
        errors.append("title: required non-empty string")

    payload = item.get("payload")
    if not isinstance(payload, dict):
        errors.append("payload: must be dict")

    created_at = item.get("created_at")
    if not isinstance(created_at, str) or "T" not in created_at:
        errors.append("created_at: must be ISO-like string containing 'T'")

    return errors


def run_all_tests_lesson_04():
    assert p1_even_odd(2) == "even"
    assert p1_even_odd(3) == "odd"

    assert p2_temp_label(0) == "cold"
    assert p2_temp_label(10) == "ok"
    assert p2_temp_label(25) == "ok"
    assert p2_temp_label(26) == "hot"

    assert p3_tariff_cost(0, 1.0) == 0.0
    assert p3_tariff_cost(10, 2.0) == 0.0
    assert p3_tariff_cost(11, 2.0) == 2.0
    assert p3_tariff_cost(15, 1.5) == 7.5

    assert p4_clamp(5, 0, 10) == 5
    assert p4_clamp(-1, 0, 10) == 0
    assert p4_clamp(11, 0, 10) == 10

    assert p5_validate_username("ab") != []
    assert p5_validate_username("valid_user_01") == []
    assert p5_validate_username("bad user") != []
    assert p5_validate_username("bad-user") != []

    ok_item = {"title": "Hello", "payload": {}, "created_at": "2026-02-25T10:00:00"}
    assert d1_validate_item(ok_item) == []

    bad_item = {"title": "   ", "payload": "notdict", "created_at": "2026-02-25"}
    errs = d1_validate_item(bad_item)
    assert len(errs) >= 2
    assert any("title" in e.lower() for e in errs)
    assert any("payload" in e.lower() for e in errs)

    print("✅ LESSON 04: all tests passed")
```

```python
run_all_tests_lesson_04()
```

# Урок 05. Циклы и статистика (задел под обработку данных диплома)

## Мини-теория
- Циклы: подсчёты, поиск, фильтрация — основа обработки данных.
- Краевые случаи: пустые списки, отсутствие совпадений.
- Диплом: первая статистика по сущностям items.

## Практика
> В этой версии вставлено **готовое решение**. Можно удалить решения и оставить TODO для студентов.


```python

# =========================
# LESSON 05 — SOLUTION
# =========================

def p1_sum_1_to_n(n: int) -> int:
    total = 0
    for i in range(1, n + 1):
        total += i
    return total


def p2_divisors_count(n: int) -> int:
    count = 0
    i = 1
    while i * i <= n:
        if n % i == 0:
            count += 1
            if i * i != n:
                count += 1
        i += 1
    return count


def p3_sign_counts(nums: list[int]) -> dict:
    pos = neg = zero = 0
    for x in nums:
        if x > 0:
            pos += 1
        elif x < 0:
            neg += 1
        else:
            zero += 1
    return {"pos": pos, "neg": neg, "zero": zero}


def p4_find_first_gt(nums: list[int], x: int) -> int:
    for i, val in enumerate(nums):
        if val > x:
            return i
    return -1


def p5_squares_of_evens(nums: list[int]) -> list[int]:
    out: list[int] = []
    for x in nums:
        if x % 2 == 0:
            out.append(x * x)
    return out


def d1_items_stats(items: list[dict]) -> dict:
    total = len(items)
    titles = set()
    last_created_at = None

    for it in items:
        title = it.get("title", "")
        if isinstance(title, str):
            t = title.strip()
            if t != "":
                titles.add(t)

        ca = it.get("created_at")
        if isinstance(ca, str):
            if last_created_at is None or ca > last_created_at:
                last_created_at = ca

    return {"total": total, "unique_titles": len(titles), "last_created_at": last_created_at}


def run_all_tests_lesson_05():
    assert p1_sum_1_to_n(0) == 0
    assert p1_sum_1_to_n(1) == 1
    assert p1_sum_1_to_n(5) == 15

    assert p2_divisors_count(1) == 1
    assert p2_divisors_count(6) == 4
    assert p2_divisors_count(10) == 4

    assert p3_sign_counts([1, -2, 0, 5, 0, -1]) == {"pos": 2, "neg": 2, "zero": 2}
    assert p3_sign_counts([]) == {"pos": 0, "neg": 0, "zero": 0}

    assert p4_find_first_gt([1, 2, 3, 10], 3) == 3
    assert p4_find_first_gt([1, 2, 3], 100) == -1
    assert p4_find_first_gt([], 0) == -1

    assert p5_squares_of_evens([1, 2, 3, 4]) == [4, 16]
    assert p5_squares_of_evens([1, 3, 5]) == []
    assert p5_squares_of_evens([0, 2]) == [0, 4]

    items = [
        {"title": " A ", "created_at": "2026-02-25T10:00:00"},
        {"title": "B", "created_at": "2026-02-25T12:00:00"},
        {"title": "A", "created_at": "2026-02-25T11:00:00"},
        {"title": "   ", "created_at": "2026-02-25T13:00:00"},
    ]
    st = d1_items_stats(items)
    assert st["total"] == 4
    assert st["unique_titles"] == 2
    assert st["last_created_at"] == "2026-02-25T13:00:00"

    st2 = d1_items_stats([])
    assert st2 == {"total": 0, "unique_titles": 0, "last_created_at": None}

    print("✅ LESSON 05: all tests passed")
```
